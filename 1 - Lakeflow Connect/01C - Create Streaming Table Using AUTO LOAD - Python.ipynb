# schemaLocation
- **What it is**: A directory path where Spark stores the inferred or explicitly defined structure (schema) of the files it reads.
- **Why it's used**: Streaming queries require a fixed schema before processing. When using tools like Auto Loader (cloudFiles), Spark uses this location to store the schema it infers from the first batch of files. If new files have altered schemas later, it handles schema evolution based on this stored metadata.

# checkpointLocation
- **What it is**: A directory path in cloud storage where the streaming query saves all of its progressive state information and offsets.
- **Why it's used**: This is the backbone of fault tolerance and exactly-once processing. If your Databricks cluster restarts, or the streaming job crashes, Spark reads this checkpoint directory upon restart to know exactly where it left off, preventing data loss or duplication.

```python
Checkpoint_file_location = "/Volumes/workspace/dbacademy/checkpoint_location/auto_loader_python"

(spark
 .readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option("delimiter", "|")
    .option("inferSchema", "true")
    .option("cloudFiles.schemaLocation", f"{Checkpoint_file_location}/schema")
    .load("/Volumes/workspace/dbacademy/samples/sales-csv/")
.writeStream
    .format("delta")
    .option("checkpointLocation", f"{Checkpoint_file_location}")
    .trigger(once=True)
    .toTable("dbacademy.sales_csv")
)
```
