```python
%sql
CREATE TABLE dbacademy.main_users_target AS
SELECT * 
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/main_users_tagret.csv',
    format => 'csv'
)
```

```python
%sql
CREATE TABLE dbacademy.main_users_source AS
SELECT * 
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/main_users_source.csv',
    format => 'csv'
)
```

```python
%sql
SELECT * FROM dbacademy.main_users_source
```

```python
%sql
select * from dbacademy.main_users_target
```

```python
%sql
MERGE INTO dbacademy.main_users_target t
USING dbacademy.main_users_source s
ON t.id = s.id
WHEN MATCHED AND s.status = 'update' THEN
  UPDATE SET 
    t.first_name = s.first_name,
    t.last_name = s.last_name,
    t.email = s.email,
    t.gender = s.gender,
    t.address = s.address,
    t.status = s.status
WHEN MATCHED AND s.status = 'delete' THEN
  DELETE
WHEN NOT MATCHED THEN
  INSERT (t.id, t.first_name, t.last_name, t.email, t.gender, t.address, t.status)
  VALUES (s.id, s.first_name, s.last_name, s.email, s.gender, s.address, s.status);

SELECT * FROM dbacademy.main_users_target
```

```python
%sql
DESCRIBE HISTORY dbacademy.main_users_target;
```

# Schema Enforcement and Schema Evolution with MERGE INTO

```python
%sql
CREATE TABLE dbacademy.new_users_source AS
SELECT * 
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/new_users_source.csv',
    format => 'csv'
)
```

```python
%sql
MERGE INTO dbacademy.main_users_target t
USING dbacademy.new_users_source s
ON t.id = s.id
WHEN MATCHED AND s.status = 'update' THEN
  UPDATE SET 
    t.first_name = s.first_name,
    t.last_name = s.last_name,
    t.email = s.email,
    t.gender = s.gender,
    t.address = s.address,
    t.status = s.status
WHEN MATCHED AND s.status = 'delete' THEN
  DELETE
WHEN NOT MATCHED THEN
  INSERT (t.id, t.first_name, t.last_name, t.email, t.gender, t.address, t.status, t.country)
  VALUES (s.id, s.first_name, s.last_name, s.email, s.gender, s.address, s.status, s.country);
```

```python
%sql
MERGE WITH SCHEMA EVOLUTION INTO dbacademy.main_users_target t
USING dbacademy.new_users_source s
ON t.id = s.id
WHEN MATCHED AND s.status = 'update' THEN
  UPDATE SET 
    t.first_name = s.first_name,
    t.last_name = s.last_name,
    t.email = s.email,
    t.gender = s.gender,
    t.address = s.address,
    t.status = s.status
WHEN MATCHED AND s.status = 'delete' THEN
  DELETE
WHEN NOT MATCHED THEN
  INSERT (id, first_name, last_name, email, gender, address, status, country)
  VALUES (s.id, s.first_name, s.last_name, s.email, s.gender, s.address, s.status, s.country);
```
