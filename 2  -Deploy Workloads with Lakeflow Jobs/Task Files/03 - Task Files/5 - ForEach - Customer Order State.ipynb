```python
df = spark.sql(f"""
               SELECT *,
                    TO_DATE(FROM_UNIXTIME(order_datetime)) as order_date,
                    CASE WHEN number_of_line_items > 2 THEN true ELSE false END as Is_large_order
               FROM dbacademy.customers_orders_silver
               WHERE state = '{state}'
               """)

table_name = f"dbacademy.customers_orders_{state.replace(" ","_").lower()}_silver"
df.write.mode("overwrite").saveAsTable(table_name)
```
