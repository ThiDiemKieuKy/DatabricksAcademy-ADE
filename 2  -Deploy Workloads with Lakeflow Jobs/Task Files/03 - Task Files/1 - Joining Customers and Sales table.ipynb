```python
%sql
DROP TABLE IF EXISTS dbacademy.customers_sales_silver;

CREATE TABLE dbacademy.customers_sales_silver AS
SELECT c.customer_id,
        c.customer_name,
        c.loyalty_segment,
        c.units_purchased,
        s.product_name,
        s.order_date,
        s.total_price
FROM dbacademy.customers_bronze c
INNER JOIN dbacademy.sales_bronze s ON c.customer_id =  s.customer_id
```

```python
df = spark.sql(f"select * from dbacademy.customers_sales_silver")

# set duplicate flag
duplicates_exist = df.count() > df.dropDuplicates().count()

# drop duplicates
df_drop_duplicates = df.drop_duplicates()

# set the boolean flag in task value
dbutils.jobs.taskValues.set("has_duplicates", duplicates_exist)
```
