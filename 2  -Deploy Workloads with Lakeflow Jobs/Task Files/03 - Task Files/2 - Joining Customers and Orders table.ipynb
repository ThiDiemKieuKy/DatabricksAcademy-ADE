```python
%sql
DROP TABLE IF EXISTS dbacademy.customers_orders_silver;

CREATE TABLE dbacademy.customers_orders_silver AS
SELECT c.customer_id,
    c.customer_name,
    c.state,
    c.city,
    o.order_number,
    o.order_datetime,
    o.number_of_line_items,
    o.promo_info
FROM dbacademy.customers_bronze c
INNER JOIN dbacademy.sales_orders_bronze o ON o.customer_id = c.customer_id
```
