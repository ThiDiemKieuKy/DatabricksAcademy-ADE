```python
# Store values from the text widgets to variables
my_catalog = dbutils.widgets.get("catalog")
my_schema =  dbutils.widgets.get("schema")
```

```python
# Set file_path
#my_catalog = 'workspace'
#my_schema = 'dbacademy'
my_volume_path = f"/Volumes/{my_catalog}/{my_schema}/trigger_storage_location/"

customer_df = (spark
               .read
               .format("csv")
               .option("header", "true")
               .option("sep", ",")
               .option("escape", "\"")
               .load(my_volume_path)
               )

customer_df.write.mode("overwrite").saveAsTable(f"{my_catalog}.{my_schema}.sales_bronze")

#display(customer_df)
```
