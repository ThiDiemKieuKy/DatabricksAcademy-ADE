```python
%sql
SELECT 
    *,
    cast(from_unixtime(user_first_touch_timestamp/1000000) as DATE) as first_touch_date,
    _metadata.file_name as file_name,
    _metadata.file_modification_time AS last_modified,
    current_timestamp() AS ingestion_time
FROM read_files('/Volumes/workspace/dbacademy/samples/users-historical/',
format => 'parquet')
LIMIT 5;
```

```python
%sql
DROP TABLE IF EXISTS users_historical_bronze;
    
CREATE TABLE users_historical_bronze AS
SELECT 
    *,
    cast(from_unixtime(user_first_touch_timestamp/1000000) as DATE) as first_touch_date,
    _metadata.file_name as file_name,
    _metadata.file_modification_time AS last_modified,
    current_timestamp() AS ingestion_time
FROM read_files('/Volumes/workspace/dbacademy/samples/users-historical/',
format => 'parquet');

SELECT * FROM users_historical_bronze LIMIT 10;
```

```python
%sql
SELECT file_name, count(1) as total_file
FROM users_historical_bronze
GROUP BY file_name
```
