```python
%sql
SELECT *
FROM csv.`/Volumes/workspace/dbacademy/samples/sales-csv/`
LIMIT 5;
```

```python
%sql
DROP TABLE IF EXISTS sales_csv_secured_data_sql;

CREATE TABLE sales_csv_secured_data_sql AS
SELECT 
    *,
    _metadata.file_name as file_name,
    _metadata.file_modification_time as last_modified_time,
    current_timestamp() as ingestion_time
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/sales-csv/',
    format => 'csv',
    header => true,
    sep => '|',
    inferSchema => true
    );
    
```

```python
%sql
SELECT * FROM sales_csv_secured_data_sql LIMIT 5;
```

```python
%sql
DESCRIBE EXTENDED sales_csv_secured_data_sql;
```

# Defining a Schema During Ingestion

```python
%sql
SELECT * 
FROM csv.`/Volumes/workspace/dbacademy/samples/malformed_example_1_data.csv`
```

```python
%sql
-- no schema, the transactions_timestamp column will be defined as a string
SELECT * 
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/malformed_example_1_data.csv',
    format => 'csv',
    header => true,
    sep => '|',
    inferSchema => true
)
```

```python
%sql
-- no schema, the transactions_timestamp column will be defined as a string
SELECT * 
FROM read_files(
    '/Volumes/workspace/dbacademy/samples/malformed_example_1_data.csv',
    format => 'csv',
    header => true,
    sep => '|',
    schema => '''
        order_id int,
        email string,
        transactions_timestamp bigint
    ''',
    rescueddatacolumn => '_rescued_data'
);
```
