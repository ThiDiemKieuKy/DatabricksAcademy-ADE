```python
%sql
CREATE OR REPLACE TABLE dbacademy.orders_bronze AS
SELECT *,
    current_timestamp() AS processing_time,
    _metadata.file_name as source_file
FROM read_files(
    '/Volumes/workspace/dbacademy/retail_pipeline/orders/',
    format => 'json'
);

CREATE OR REPLACE TABLE dbacademy.orders_silver AS
SELECT order_id,
        timestamp(order_timestamp) as order_timestamp,
        customer_id,
        notifications
FROM dbacademy.orders_bronze;

CREATE OR REPLACE TABLE dbacademy.orders_by_date_view AS
SELECT date(order_timestamp) as order_date,
        count(*) as total_daily_orders
FROM dbacademy.orders_silver
GROUP BY date(order_timestamp)
```
